```python
df = (
    watchhistory
    .merge(movies, on="movie_id", how="left")
    .merge(users, on="user_id", how="left")
)

print("Joined dataframe:", df.shape)
display(df.head(10))
```

```python
print("Genres in catalog:")
display(movies["genres"].value_counts())

print("\nMost watched genres:")
display(df["genres"].value_counts())
```

```python
movie_ratings = (
    df.groupby(["movie_id", "Title", "genres", "release_year"])
    .agg(
        avg_rating=("Rating", "mean"),
        num_ratings=("Rating", "count")
    )
    .reset_index()
    .sort_values("avg_rating", ascending=False)
)

print("Movies ranked by average rating:")
display(movie_ratings)
```

```python
user_activity = (
    df.groupby(["user_id", "name"])
    .agg(
        movies_watched=("movie_id", "count"),
        avg_rating_given=("Rating", "mean"),
        first_watch=("watch_date", "min"),
        last_watch=("watch_date", "max")
    )
    .reset_index()
    .sort_values("movies_watched", ascending=False)
)

print("User activity:")
display(user_activity)
```

```python
df["year_month"] = df["watch_date"].dt.to_period("M")

monthly = (
    df.groupby("year_month")
    .agg(
        avg_rating=("Rating", "mean"),
        num_watches=("Rating", "count")
    )
    .reset_index()
)
monthly["year_month"] = monthly["year_month"].astype(str)

print("Average rating and number of watches per month:")
display(monthly)
```

```python
user_genre = (
    df.groupby(["user_id", "name", "genres"])
    .size()
    .reset_index(name="count")
    .sort_values(["user_id", "count"], ascending=[True, False])
)

favorite_genre = (
    user_genre
    .groupby(["user_id", "name"])
    .first()
    .reset_index()
    .rename(columns={"genres": "favorite_genre", "count": "times_watched"})
)

print("Most watched genre per user:")
display(favorite_genre)
```
